1A.1. Show that $\alpha + \beta = \beta + \alpha$ for all $\alpha, \beta \in \mathbb{C}$.

Since $\alpha = a + bi$ and $\beta = c + di$, we have that

  $\alpha + \beta = a + bi + c + di = a + c + bi + di =$

  $= c + di + a + bi = \beta + \alpha$
  
  $\square$

1A.2. Show that $(\alpha + \beta) + \lambda = \alpha + (\beta + \lambda)$ for all $\alpha, \beta, \lambda \in \mathbb{C}$.

Since $\alpha = a + bi, \beta = c + di, \lambda = e + fi$, we have that

$(\alpha + \beta) + \lambda = (a + bi + c + di) + e + fi =$

$= a + bi + c + di + e + fi = a + bi + (c + di + e + fi) = \alpha + (\beta + \lambda)$

$\square$

1A.3. Show that $(\alpha\beta)\lambda = \alpha(\beta\lambda)$ for all $\alpha, \beta, \lambda \in \mathbb{C}.$

Since $\alpha = a + bi, \beta = c + di, \lambda = e + fi$, we have that

$(\alpha\beta)\lambda = ((a + bi)(c + di))(e + fi) =$

$= ((ac - bd) + (ad + bc)i)(e + fi) =$

$= (ace - bde - adf - bcf) + (acf - bdf + ade + bce)i =$

$= (ace - adf - bde - bcf) + (ade + acf + bce - bdf)i =$

$= (a + bi)((ce - df) + (de + cf)i) =$

$= (a + bi)((c + di)(e + fi)) = \alpha(\beta\lambda)$

$\square$

1A.4. Show that $\lambda(\alpha + \beta) = \lambda\alpha + \lambda\beta$ for all $\lambda, \alpha, \beta \in \mathbb{C}$.

Since $\lambda = a + bi, \alpha = c + di, \beta = e + fi$, we have that

$\lambda(\alpha + \beta) = (a + bi)(c + di + e + fi) =$

$= ac + adi + ae + afi + bci - bd + bei - bf =$

$= (ac + adi + bci - bd) + (ae + afi + bei - bf) =$

$= (a + bi)(c + di) + (a + bi)(e + fi) =$

$= \lambda\alpha + \lambda\beta$

$\square$

1A.5. Show that for every $\alpha \in \mathbb{C}$, there exists an unique $\beta \in \mathbb{C}$ such that $\alpha + \beta = 0$.

Let's define $\beta = x + yi$, with $x, y \in \mathbb{R}$ as any complex number such that $\alpha + \beta = 0$.

We have that $\alpha + \beta = a + bi + x + yi = (a + x) + (b + y)i = 0 + 0i$.

From this we obtain that $x = -a$ and $y = -b$, which are both unique unknown constants.

$\square$

1A.6. Show that for every $\alpha \in \mathbb{C}$ with $\alpha \neq 0$, there exists an unique $\beta \in \mathbb{C}$ such that $\alpha\beta = 1$.

Let's define $\beta = x + yi$, with $x, y \in \mathbb{R}$ as any complex number such that $\alpha\beta = 1$.

We have that $\alpha\beta = (a + bi)(x + yi) = ax + ayi + bxi - by =$

$= (ax - by) + (ay + bx)i = 1 + 0i$.

From this we have the equations system

$$
  ax - by = 1\\
  bx + ay = 0
$$
$$
  aax - aby = a\\
  bbx + aby = b
$$
$$
  (aa + bb)x = a + b
$$
Since we know $a$ and $b$ cannot be $0$ at the same time
$$
  x = {{a+b}\over{a^2+b^2}}
$$
$$
  y = -{{bx}\over{a}}\text{, if } a \neq 0
$$
$$
  y = -{{ab+b^2}\over{a^3+ab^2}}
$$
If $a = 0$
$$
x = {1\over{b}}
$$
$$
  {{a}\over{b}} - by = 1
$$
Since $b \neq 0$
$$
  y = -{1\over{b}}
$$
Being all values of $x$ and $y$ unknown constants.

$\square$

1A.7. Show that

$${-1 + \sqrt{3}i}\over{2}$$

is a cube root of $1$ (meaning that its cube equals $1$).

For this, we simply have to multiply the rational by itself and then again:

$$
\left({{-1 + \sqrt{3}i}\over{2}}\right)\left({{-1 + \sqrt{3}i}\over{2}}\right)\left({{-1 + \sqrt{3}i}\over{2}}\right)=\\
=\left({{1 - 2\sqrt{3}i - 3}\over{4}}\right)\left({{-1 + \sqrt{3}i}\over{2}}\right)=\\
=\left({{-1-\sqrt{3}i}\over{2}}\right)\left({{-1 + \sqrt{3}i}\over{2}}\right)=\\
={{1-\sqrt{3}i+\sqrt{3}i+3}\over{4}}=1
$$

$\square$

1A.11. Show that $(x + y) + z = x + (y + z)$ for all $x,y,z\in\mathbb{F}^n$.

We have that

$$
(x + y) + z = \left((x_1,...,x_n) + (y_1,...,y_n)\right)+(z_1,...,z_n)=\\
=(x_1+y_1,...,x_n+y_n)+(z_1,...,z_n)=\\
=(x_1+y_1+z_1,...,x_n+y_n+z_n)=\\
=(x_1,...,z_1)+(y_1+z_1,...,y_n+z_n)=\
=(x_1,...,z_1)+\left((y_1,...,y_n)+(z_1,...,z_n)\right)=\\
=x+(y+z)\\
\square
$$

1A.14. Show that $\lambda(x+y)=\lambda x + \lambda y$ for all $\lambda\in\mathbb{F}$ and all $x,y\in\mathbb{F}^n$.

If $S$ is a nonempty set, then $\mathbb{F}^S$ is a vector space over $\mathbb{F}.$

**Commutativity**

Let $u,v\in\mathbb{F}^S$.

$u + v = (u+v)(x), x\in S$

$(u + v)(x) = u(x) + v(x)$

Since $u(x), v(x)\in \mathbb{F}$:

$u(x) + v(x) = v(x) + u(x) = (v + u)x = v + u$ $\square$

**Additive associativity**

Let $u,v,w\in\mathbb{F}^S$

$(u + v) + w = (u + v)(x) + w(x) = u(x) + v(x) + w(x), x\in S$

Since $u(x), v(x), w(x)\in \mathbb{F}$:

$u(x) + v(x) + w(x) = u(x) + (v(x) + w(x)) = u + (v + w)$ $\square$

**Multiplicative associativity**

Let $a,b\in\mathbb{F},v\in\mathbb{F}^S$

$(ab)v = (ab)v(x), x\in S$

Since $a, b, v(x)\in \mathbb{F}$:

$(ab)v(x) = a(bv(x)) = a(bv)$ $\square$
