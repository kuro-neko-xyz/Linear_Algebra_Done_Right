1A.1. Show that $\alpha + \beta = \beta + \alpha$ for all $\alpha, \beta \in \mathbb{C}$.

Since $\alpha = a + bi$ and $\beta = c + di$, we have that

  $\alpha + \beta = a + bi + c + di = a + c + bi + di =$

  $= c + di + a + bi = \beta + \alpha$
  
  $\square$

1A.2. Show that $(\alpha + \beta) + \lambda = \alpha + (\beta + \lambda)$ for all $\alpha, \beta, \lambda \in \mathbb{C}$.

Since $\alpha = a + bi, \beta = c + di, \lambda = e + fi$, we have that

$(\alpha + \beta) + \lambda = (a + bi + c + di) + e + fi =$

$= a + bi + c + di + e + fi = a + bi + (c + di + e + fi) = \alpha + (\beta + \lambda)$

$\square$

1A.3. Show that $(\alpha\beta)\lambda = \alpha(\beta\lambda)$ for all $\alpha, \beta, \lambda \in \mathbb{C}.$

Since $\alpha = a + bi, \beta = c + di, \lambda = e + fi$, we have that

$(\alpha\beta)\lambda = ((a + bi)(c + di))(e + fi) =$

$= ((ac - bd) + (ad + bc)i)(e + fi) =$

$= (ace - bde - adf - bcf) + (acf - bdf + ade + bce)i =$

$= (ace - adf - bde - bcf) + (ade + acf + bce - bdf)i =$

$= (a + bi)((ce - df) + (de + cf)i) =$

$= (a + bi)((c + di)(e + fi)) = \alpha(\beta\lambda)$

$\square$

1A.4. Show that $\lambda(\alpha + \beta) = \lambda\alpha + \lambda\beta$ for all $\lambda, \alpha, \beta \in \mathbb{C}$.

Since $\lambda = a + bi, \alpha = c + di, \beta = e + fi$, we have that

$\lambda(\alpha + \beta) = (a + bi)(c + di + e + fi) =$

$= ac + adi + ae + afi + bci - bd + bei - bf =$

$= (ac + adi + bci - bd) + (ae + afi + bei - bf) =$

$= (a + bi)(c + di) + (a + bi)(e + fi) =$

$= \lambda\alpha + \lambda\beta$

$\square$

1A.5. Show that for every $\alpha \in \mathbb{C}$, there exists an unique $\beta \in \mathbb{C}$ such that $\alpha + \beta = 0$.

Since $\alpha = a + bi, \beta = c + di$, we have that

$\alpha + \beta = (a + bi) + (c + di)$.

Therefore, we have $(a + c) + (b + d)i = 0 + 0i$.

By the additive inverse of real numbers, There is only one $c \in \mathbb{R}$ such that $a + c = 0$. Conversely, there is only one $d \in \mathbb{R}$ such that $b + d = 0$.

As such, the complex number defined as $c + di$ is unique.

$\square$

1A.6. Show that for every $\alpha \in \mathbb{C}$ with $\alpha \neq 0$, there exists an unique $\beta \in \mathbb{C}$ such that $\alpha\beta = 1$.

Let's define $\beta = x + yi$, with $x, y \in \mathbb{R}$ as any complex number such that $\alpha\beta = 1$.

We have that $\alpha\beta = (a + bi)(x + yi) = ax + ayi + bxi - by =$

$= (ax - by) + (ay + bx)i = 1 + 0i$.

From this we have the equations system

$$
  ax - by = 1\\
  bx + ay = 0
$$
$$
  aax - aby = a\\
  bbx + aby = b
$$
$$
  (aa + bb)x = a + b
$$
Since we know $a$ and $b$ cannot be $0$ at the same time
$$
  x = {{a+b}\over{a^2+b^2}}
$$
$$
  y = -{{bx}\over{a}}\text{, if } a \neq 0
$$
$$
  y = -{{ab+b^2}\over{a^3+ab^2}}
$$
If $a = 0$
$$
x = {1\over{b}}
$$
$$
  {{a}\over{b}} - by = 1
$$
Since $b \neq 0$
$$
  y = -{1\over{b}}
$$
Being all values of $x$ and $y$ unknown constants.

$\square$
